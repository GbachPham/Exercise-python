### Step 1. Import the necessary libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

# set this so the graphs open internally
%matplotlib inline
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/chipotle.tsv).

### Step 3. Assign it to a variable called chipo.

```python
url = 'https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/chipotle.tsv'
chipo = pd.read_csv(url, sep = '\t')
chipo.head()
```

### Step 4. See the first 10 entries

```python
chipo.head(10)
```

### Step 5. Create a histogram of the top 5 items bought

```python
top_5 = chipo.groupby('item_name').quantity.sum().sort_values(ascending=False).head(5)
top_5.plot(kind='bar')
plt.xlabel('Items')
plt.ylabel('Number of Times Ordered')
plt.title('Most ordered Chipotle\'s Items')
plt.show()
```

### Step 6. Create a scatterplot with the number of items orderered per order price
#### Hint: Price should be in the X-axis and Items ordered in the Y-axis

```python
if chipo['item_price'].dtype == object:
    chipo['item_price'] = chipo['item_price'].apply(lambda x: float(x[1:-1]))
orders = chipo.groupby('order_id').sum(numeric_only=True)
plt.scatter(x = orders.item_price, y = orders.quantity, s = 50, c = 'blue')
plt.xlabel('Order Price')
plt.ylabel('Items ordered')
plt.title('Number of items ordered per order price')
plt.ylim(0)
plt.show()
```

### Step 7. BONUS: Create a question and a graph to answer your own question.

```python
chipo['item_price'].plot(kind='hist', bins=20)
plt.xlabel('Price ($)')
plt.title('Chipotle Item Price Distribution')
plt.show()
```
