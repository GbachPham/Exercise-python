# Visualizing the Titanic Disaster

### Introduction:

This exercise is based on the titanic Disaster dataset avaiable at [Kaggle](https://www.kaggle.com/c/titanic).  
To know more about the variables check [here](https://www.kaggle.com/c/titanic/data)


### Step 1. Import the necessary libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/titanic_train.csv)

### Step 3. Assign it to a variable titanic 

```python
titanic = pd.read_csv('https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/titanic_train.csv')
```

### Step 4. Set PassengerId as the index 

```python
titanic = titanic.set_index('PassengerId')
```

### Step 5. Create a pie chart presenting the male/female proportion

```python
titanic['Sex'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.ylabel('')
plt.title('Male/Female Proportion on Titanic')
plt.show()
```

### Step 6. Create a scatterplot with the Fare payed and the Age, differ the plot color by gender

```python
sns.scatterplot(x='Age', y='Fare', hue='Sex', data=titanic, alpha=0.6)
plt.title('Age vs Fare by Gender')
plt.show()
```

### Step 7. How many people survived?

```python
titanic['Survived'].value_counts()
print(f"\nTotal survived: {titanic['Survived'].sum()}")
print(f"Total did not survive: {(titanic['Survived'] == 0).sum()}")
```

### Step 8. Create a histogram with the Fare payed

```python
titanic['Fare'].hist(bins=30)
plt.xlabel('Fare')
plt.ylabel('Frequency')
plt.title('Distribution of Fare Paid')
plt.show()
```

### BONUS: Create your own question and answer it.

```python
# Question: What is the survival rate by passenger class?
sns.barplot(x='Pclass', y='Survived', data=titanic)
plt.ylabel('Survival Rate')
plt.xlabel('Passenger Class')
plt.title('Survival Rate by Passenger Class')
plt.show()
```
