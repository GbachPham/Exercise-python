### Step 1. Import the necessary libraries:

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/tips.csv).

### Step 3. Assign it to a variable called tips

```python
tips = pd.read_csv('https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/tips.csv')
```

### Step 4. Delete the Unnamed 0 column

```python
tips = tips.drop('Unnamed: 0', axis=1)
```

### Step 5. Plot the total_bill column histogram

```python
tips['total_bill'].hist(bins=20)
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.title('Distribution of Total Bill')
plt.show()
```

### Step 6. Create a scatter plot presenting the relationship between total_bill and tip

```python
plt.scatter(tips['total_bill'], tips['tip'], alpha=0.5)
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.title('Total Bill vs Tip')
plt.show()
```

### Step 7.  Create one image with the relationship of total_bill, tip and size.
#### Hint: It is just one function.

```python
sns.relplot(x='total_bill', y='tip', size='size', data=tips)
```

### Step 8. Present the relationship between days and total_bill value

```python
sns.boxplot(x='day', y='total_bill', data=tips)
```

### Step 9. Create a scatter plot with the day as the y-axis and tip as the x-axis, differ the dots by sex

```python
sns.stripplot(x='tip', y='day', hue='sex', data=tips)
```

### Step 10.  Create a box plot presenting the total_bill per day differetiation the time (Dinner or Lunch)

```python
sns.boxplot(x='day', y='total_bill', hue='time', data=tips)
```

### Step 11. Create two histograms of the tip value based for Dinner and Lunch. They must be side by side.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
tips[tips['time'] == 'Lunch']['tip'].hist(bins=15, ax=axes[0])
axes[0].set_title('Lunch Tips')
axes[0].set_xlabel('Tip')
tips[tips['time'] == 'Dinner']['tip'].hist(bins=15, ax=axes[1])
axes[1].set_title('Dinner Tips')
axes[1].set_xlabel('Tip')
plt.show()
```

### Step 12. Create two scatterplots graphs, one for Male and another for Female, presenting the total_bill value and tip relationship, differing by smoker or no smoker
### They must be side by side.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
male_data = tips[tips['sex'] == 'Male']
female_data = tips[tips['sex'] == 'Female']

for smoker in male_data['smoker'].unique():
    male_smoker = male_data[male_data['smoker'] == smoker]
    axes[0].scatter(male_smoker['total_bill'], male_smoker['tip'], label=f'Male {smoker}', alpha=0.6)
axes[0].set_xlabel('Total Bill')
axes[0].set_ylabel('Tip')
axes[0].set_title('Male: Total Bill vs Tip')
axes[0].legend()

for smoker in female_data['smoker'].unique():
    female_smoker = female_data[female_data['smoker'] == smoker]
    axes[1].scatter(female_smoker['total_bill'], female_smoker['tip'], label=f'Female {smoker}', alpha=0.6)
axes[1].set_xlabel('Total Bill')
axes[1].set_ylabel('Tip')
axes[1].set_title('Female: Total Bill vs Tip')
axes[1].legend()

plt.tight_layout()
plt.show()
```

### BONUS: Create your own question and answer it using a graph.

```python
# Question: What is the tip percentage by day of week?
tips['tip_percentage'] = (tips['tip'] / tips['total_bill']) * 100
sns.boxplot(x='day', y='tip_percentage', data=tips)
plt.ylabel('Tip Percentage (%)')
plt.title('Tip Percentage Distribution by Day of Week')
plt.show()
```
