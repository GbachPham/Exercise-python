### Step 1. Import the necessary libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

### Step 2. Create the DataFrame that should look like the one below.

```python
raw_data = {'age': [24, 24, 24, 25, 25, 25, 25, 26, 26, 26, 26, 26, 26, 27, 27, 27, 27, 27, 28, 28, 28, 28, 28, 28, 28, 29, 29, 29, 29, 29],
        'preTestScore': [40, 50, 60, 30, 40, 50, 60, 50, 60, 70, 50, 60, 70, 50, 60, 70, 40, 50, 50, 60, 70, 40, 50, 60, 70, 50, 60, 70, 45, 55],
        'postTestScore': [50, 62, 75, 45, 52, 67, 78, 65, 78, 82, 60, 75, 80, 65, 75, 85, 55, 65, 60, 75, 85, 50, 60, 70, 80, 65, 75, 85, 60, 70],
        'sex': ['M', 'F', 'M', 'F', 'M', 'F', 'M', 'F', 'M', 'F', 'M', 'F', 'M', 'F', 'M', 'F', 'M', 'F', 'M', 'F', 'M', 'F', 'M', 'F', 'M', 'F', 'M', 'F', 'M', 'F']}

df = pd.DataFrame(raw_data)
df
```

### Step 3. Create a Scatterplot of preTestScore and postTestScore, with the size of each point determined by age
#### Hint: Don't forget to place the labels

```python
plt.scatter(df['preTestScore'], df['postTestScore'], s=df['age']*10, alpha=0.5)
plt.xlabel('preTestScore')
plt.ylabel('postTestScore')
plt.title('preTestScore vs postTestScore (size by age)')
plt.show()
```

### Step 4. Create a Scatterplot of preTestScore and postTestScore.
### This time the size should be 4.5 times the postTestScore and the color determined by sex

```python
plt.scatter(df['preTestScore'], df['postTestScore'], s=df['postTestScore']*4.5, c=df['sex'].map({'M': 1, 'F': 0}), alpha=0.5, cmap='Set1')
plt.xlabel('preTestScore')
plt.ylabel('postTestScore')
plt.title('preTestScore vs postTestScore (size by postTestScore, color by sex)')
plt.show()
```

### BONUS: Create your own question and answer it.

```python
# Question: What is the relationship between age and improvement (postTestScore - preTestScore)?
df['improvement'] = df['postTestScore'] - df['preTestScore']
plt.scatter(df['age'], df['improvement'], s=50, alpha=0.6, c=df['sex'].map({'M': 'blue', 'F': 'red'}))
plt.xlabel('Age')
plt.ylabel('Improvement (postTestScore - preTestScore)')
plt.title('Improvement by Age (Blue=Male, Red=Female)')
plt.legend(['Male', 'Female'])
plt.show()
```
